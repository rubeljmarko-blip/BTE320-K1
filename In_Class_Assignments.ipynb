## In Class Assignment 1 - Income Statement

```python
# Inputs:
sales = float(input("Enter sales ($): "))
tax_rate = float(input("Enter tax rate (%): "))

# Processes:
cogs = sales * 0.40
gross_profit = sales - cogs
taxes = gross_profit * (tax_rate / 100)
net_income = gross_profit - taxes

# Outputs:
print(f"Sales: $ {sales:,.2f}")
print(f"Cost of Goods Sold: $ {cogs:,.2f}")
print(f"Gross Profit: $ {gross_profit:,.2f}")
print(f"Taxes: $ {taxes:,.2f}")
print(f"Net Income: $ {net_income:,.2f}")
```

## In Class Assignment 2 - Numerical Calculator

```python
# Inputs:
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))
operator = input("Enter the operator (+, -, *, /): ")

# Processes:
if operator == "+":
    result = num1 + num2
elif operator == "-":
    result = num1 - num2
elif operator == "*":
    result = num1 * num2
elif operator == "/":
    if num2 != 0:
        result = num1 / num2
    else:
        result = "Error: Division by zero"
else:
    result = "Error: Invalid operator"

# Outputs:
print(f"{num1} {operator} {num2} = {result}")
```

## In Class Assignment 5 - Movie Theater

```python
# Movie theater advertising vs profit

# Processes & Outputs:
ticket_price = 10
base_attendees = 20
fixed_costs = 200

print("Ad$", "Profit")

ad_spend = 0
while ad_spend <= 200:
    additional = 2 * (ad_spend ** 0.5)
    total_attendees = base_attendees + additional
    revenue = total_attendees * ticket_price
    profit = revenue - fixed_costs - ad_spend
    print(ad_spend, round(profit, 1))
    ad_spend += 25
```

## In Class Assignment 6 - Rock Paper Scissors

```python
import random

# Processes:
play = input("Do you want to play Rock-Paper-Scissors? (yes/no): ")

while play.lower() == "yes":
    # Inputs:
    user_choice = input("Enter rock, paper, or scissors: ").lower()
    computer_choice = random.choice(["rock", "paper", "scissors"])

    print(f"Computer chose: {computer_choice}")

    # Determine winner
    if user_choice == computer_choice:
        print("It's a tie!")
    elif (user_choice == "rock" and computer_choice == "scissors") or \
         (user_choice == "paper" and computer_choice == "rock") or \
         (user_choice == "scissors" and computer_choice == "paper"):
        print("You win!")
    else:
        print("You lose!")

    play = input("Do you want to play again? (yes/no): ")

# Outputs:
print("Thanks for playing!")
```

## In Class Assignment 12 - Recursive Function: String Reversal

```python
# Processes:
def reverse(s):
    if len(s) == 0:
        return s
    else:
        return reverse(s[1:]) + s[0]

# Inputs:
text = input("Enter a string: ")

# Outputs:
print(f"reverse('{text}') = '{reverse(text)}'")
```

## In Class Assignment 13 - ROI Calculator

```python
# Processes:
def roi(cost, revenue):
    if not isinstance(cost, (int, float)) or not isinstance(revenue, (int, float)):
        return "Error: Both inputs must be numeric"
    if cost <= 0:
        return "Error: Cost must be greater than 0"
    return (revenue - cost) / cost

# Inputs:
cost = float(input("Enter cost: "))
revenue = float(input("Enter revenue: "))

# Outputs:
result = roi(cost, revenue)
if isinstance(result, float):
    print(f"ROI: {result:.2f} ({result * 100:.1f}%)")
else:
    print(result)
```

## In Class Assignment 14 - Loan Payment Calculator

```python
# Processes:
def loan_payment(principal, annual_rate, years):
    if principal <= 0 or annual_rate <= 0 or years <= 0:
        return "Error: All inputs must be positive numbers"
    r = annual_rate / 100 / 12
    n = years * 12
    payment = (r * principal) / (1 - (1 + r) ** (-n))
    return payment

# Inputs:
principal = float(input("Enter loan amount ($): "))
annual_rate = float(input("Enter annual interest rate (%): "))
years = float(input("Enter loan term (years): "))

# Outputs:
result = loan_payment(principal, annual_rate, years)
if isinstance(result, float):
    print(f"Monthly payment: ${result:,.2f}")
else:
    print(result)
```

## In Class Assignment 15 - Quality Control Manager (Exceptions)

```python
# Processes:
def getValues(n):
    values = []
    count = 0
    while count < n:
        try:
            val = float(input(f"Enter value {count + 1}: "))
            if val < 0:
                print("Error: value cannot be negative. Try again.")
                continue
            values.append(val)
            count += 1
        except ValueError:
            print("Error: please enter a valid number.")
    return values

def calcAverage(alist):
    # Exclude the first three values
    trimmed = alist[3:]
    if len(trimmed) == 0:
        return "Error: not enough values to calculate an average"
    return sum(trimmed) / len(trimmed)

# Inputs:
try:
    n = int(input("How many values would you like to enter? "))
    if n <= 3:
        print("Error: you must enter more than 3 values.")
    else:
        values = getValues(n)

        # Outputs:
        result = calcAverage(values)
        if isinstance(result, float):
            print(f"Values entered: {values}")
            print(f"Average (excluding first 3): {result:.4f}")
        else:
            print(result)
except ValueError:
    print("Error: please enter a valid integer for n.")
```

## In Class Assignment 16 - Greeting Class

```python
# Processes:
class Greeting:
    def __init__(self, first, last):
        self.first = first
        self.last = last

    def display(self, formal):
        if formal:
            print(f"{self.last}, {self.first}")
        else:
            print(f"{self.first}")

    def __str__(self):
        return f"Greeting(first='{self.first}', last='{self.last}')"

# Inputs:
first = input("Enter first name: ")
last = input("Enter last name: ")

# Outputs:
person = Greeting(first, last)
print("Formal:")
person.display(True)
print("Informal:")
person.display(False)
print("__str__ output:")
print(person)
```

## In Class Assignment 17 - Student Class

```python
# Processes:
class Student:
    def __init__(self, name, number, courses=None):
        self.__name = name
        self.__number = number
        self.__courses = courses if courses is not None else []

    # Getter for name
    def get_name(self):
        return self.__name

    # Getter for courses
    def get_courses(self):
        return self.__courses

    # Setter for name
    def set_name(self, name):
        self.__name = name

    # Special method: length of courses
    def __len__(self):
        return len(self.__courses)

    # Enroll a course if not already present
    def enroll(self, course):
        if course not in self.__courses:
            self.__courses.append(course)
            print(f"Enrolled in {course}")
        else:
            print(f"Already enrolled in {course}")

# Inputs & Outputs:
s = Student("Marko Rubelj", 12345)

s.enroll("BTE320")
s.enroll("BTE101")
s.enroll("BTE320")  # duplicate -- should be skipped

print(f"Name: {s.get_name()}")
print(f"Courses: {s.get_courses()}")
print(f"Number of courses: {len(s)}")

# Demonstrate setter
s.set_name("Marko J. Rubelj")
print(f"Updated name: {s.get_name()}")
```
